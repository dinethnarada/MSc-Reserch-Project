METEOR Score

```python
from datasets import load_metric

# Load METEOR metric
meteor = load_metric("meteor")

def get_meteor_score_from_files(reference_file_path, predicted_file_path):
    # Read reference and predicted lines
    with open(reference_file_path, 'r', encoding='utf-8') as ref_file:
        references = [line.strip() for line in ref_file.readlines()]

    with open(predicted_file_path, 'r', encoding='utf-8') as pred_file:
        predictions = [line.strip() for line in pred_file.readlines()]

    # Ensure same length
    if len(references) != len(predictions):
        raise ValueError("Reference and prediction files must have the same number of lines.")

    # Compute METEOR
    result = meteor.compute(predictions=predictions, references=references)
    return round(result["meteor"], 4)

score = get_meteor_score_from_files(original, predicted)
print("METEOR Score:", score)
```
